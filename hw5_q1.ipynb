```python
import numpy as np
import scipy as sc
import matplotlib.pyplot as plt
```

```python
def system(t, state, mu, rho):
    x = state[0]
    y = state[1]
    dx_dt = (1-mu * x) * x**(1.5) - x**(0.5)*y
    dy_dt = rho*(x-1)*y
    return [dx_dt, dy_dt]
```

```python
t_start = 0
t_end = 100
t_vals = np.linspace(t_start, t_end, 50000)

mu = 0.45
rho = 50

x0 = [1.05, 1-mu]

sol = sc.integrate.solve_ivp(system, [t_start, t_end], x0, args=(mu,rho), max_step=0.001)

trans_cutoff = 0
mask = sol.t >= trans_cutoff
t_steady = sol.t[mask]
x = sol.y[0][mask]
y = sol.y[1][mask]

y_vals = lambda x: 40 * (1-mu * x)
y_vals2 = lambda x: 20 * x - 0.01

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))
ax1.plot(t_steady, x, lw=0.7)
ax1.set_title('Sawtooth Oscillations')
ax1.set_xlabel('$t$')
ax1.set_ylabel('$x(t)$')

ax2.plot(x, y)
ax2.plot(1, 1-mu, 'ko')
ax2.set_title('Phase Plane')
ax2.set_xlabel('$x$')
ax2.set_ylabel('$y$')
ax2.set_ylim([0, 40])
```

```python
mu1 = 0.25
rho1 = 20

mu2 = 0.25
rho2 = 50

mu3 = 0.45
rho3 = 20

mu4 = 0.45
rho4 = 50

x01 = [1.05, 1-mu1]
x02 = [1.05, 1-mu2]
x03 = [1.05, 1-mu3]
x04 = [1.05, 1-mu4]


sol1 = sc.integrate.solve_ivp(system, [t_start, t_end], x01, args=(mu1,rho1), max_step = 0.01)
sol2 = sc.integrate.solve_ivp(system, [t_start, t_end], x02, args=(mu2,rho2), max_step = 0.01)
sol3 = sc.integrate.solve_ivp(system, [t_start, t_end], x03, args=(mu3,rho3), max_step = 0.01)
sol4 = sc.integrate.solve_ivp(system, [t_start, t_end], x04, args=(mu4,rho4), max_step = 0.01)


fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(10, 10))
ax1.plot(sol1.t, sol1.y[0], lw=0.7)
ax1.set_title(r'$\mu = 0.25, \; \rho = 20$')
ax1.set_xlabel('$t$')
ax1.set_ylabel('$x(t)$')

ax2.plot(sol2.t, sol2.y[0], lw=0.7)
ax2.set_title(r'$\mu = 0.25, \; \rho = 50$')
ax2.set_xlabel('$t$')
ax2.set_ylabel('$x(t)$')

ax3.plot(sol3.t, sol3.y[0], lw=0.7)
ax3.set_title(r'$\mu = 0.45, \; \rho = 20$')
ax3.set_xlabel('$t$')
ax3.set_ylabel('$x(t)$')

ax4.plot(sol4.t, sol4.y[0], lw=0.7)
ax4.set_title(r'$\mu = 0.45, \; \rho = 50$')
ax4.set_xlabel('$t$')
ax4.set_ylabel('$x(t)$')
```
